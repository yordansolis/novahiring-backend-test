# NovaHiring — Informe de Selección

Este notebook llama a `GET /api/v1/jobs/{job_id}/report` y muestra el informe final en Markdown.

## Requisitos
```bash
uv run fastapi dev main.py   # servidor corriendo
uv run python seed.py        # datos sembrados
```

Ejecutar **después** de `e2e_interview.ipynb` para que el informe incluya los scores reales de la entrevista.

````python
import httpx
from IPython.display import Markdown, display

BASE_URL = 'http://localhost:8000'
JOB_ID   = 'job-clinica-salud-valencia-001'

client = httpx.Client(base_url=BASE_URL, timeout=30.0)
print('✓ Cliente HTTP listo')
````

````python
# Paso 1: verificar servidor
try:
    r = client.get('/health')
    r.raise_for_status()
    print(f'✓ Servidor OK → {r.json()}')
except httpx.ConnectError:
    print('✗ Servidor no está corriendo. Ejecuta: uv run fastapi dev main.py')
    raise SystemExit(1)
````

````python
# Paso 2: obtener el informe
r = client.get(f'/api/v1/jobs/{JOB_ID}/report')
r.raise_for_status()

report_md = r.text
print(f'✓ Informe generado ({len(report_md)} caracteres)')
````

````python
# Paso 3: renderizar el informe como Markdown
display(Markdown(report_md))
````

````python
# Paso 4: guardar el informe en disco
from pathlib import Path

output_path = Path().resolve().parent / 'docs' / 'reports' / 'ranking-interview-result.md'
output_path.write_text(report_md, encoding='utf-8')
print(f'✓ Informe guardado en: {output_path}')
````
